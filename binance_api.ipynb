```python
import json
import pandas as pd
import websocket
```

```python
crypto = 'dogeusdt'
interval = '1m'

closes = []
highs = []
lows = []
volumes = []

# documentation was used as a base for following functions

def on_message(ws, message):
    json_message = json.loads(message)
    candle = json_message['k']

    is_closed = candle['x']

    if is_closed:
        close_p = float(candle['c'])
        high_p = float(candle['h'])
        low_p = float(candle['l'])
        volume = float(candle['v'])

        closes.append(close_p)
        highs.append(high_p)
        lows.append(low_p)
        volumes.append(volume)
        print(f'Close prise is: {close_p}')

def on_close(ws, close_status_code, close_msg):
    print("### closed ###")

def on_open(ws):
    print("Opened connection")

socket = f'wss://stream.binance.com:9443/ws/{crypto.lower()}@kline_{interval}'
ws = websocket.WebSocketApp(socket, on_open=on_open, on_message=on_message, on_close=on_close)

ws.run_forever()
```
